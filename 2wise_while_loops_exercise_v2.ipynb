### Exercise 1 — Password with limited attempts
Keep asking for a password until the user enters `'python123'` **or** the maximum number of attempts is reached (read the max with `input()`). Print `'Access granted!'` if correct; otherwise tell them how many attempts remain.

Example output (max = 3, all wrong):
```
Password incorrect! You have 2 attempts left
Password incorrect! You have 1 attempts left
Password incorrect! You have 0 attempts left
```
*Tip: track a counter (or count down remaining attempts). Loop while attempts remain; `break` as soon as the password is correct.*

````python
max_attempts = int(input("How many attempts? "))
# Your code here
count = max_attempts
pwd = ""
while pwd != "python123" and count != 0:
    pwd = input("\nInput Password:")
    count -= 1
    if pwd != "python123":
        print(f'Password Incorrect! You have {count} attempts left')
if pwd == "python123":
    print("Access Granted!")
else:
    print(f'Access Denied! You have {count} attempts left')
````

### Exercise 2 — Number guessing game (while version)
Rewrite the guessing game with a `while` loop, stopping when correct **or** when the guesses run out.
- `guess == target` -> "Correct!"    - off by 1 -> "Close!"
- `guess > target + 1` -> "Too High"  - otherwise -> "Too Low"

Example output:
```
Too Low!
Too Low!
Too Low!
Close!
Correct!
```
*Tip: `while guesses_left > 0:` (decrement each pass), or `while True:` with a counter and a `break`.*

````python
import random
target = random.randint(1, 10)
# Your code here
count = 5
while True:
    if count != 0:
        guess = int(input("\nGuess a number:"))
        count -= 1
        if guess == target:
            print("Correct")
            break
        elif guess == target + 1 or guess == target - 1:
            print("Close")
            print(f'{count} Guesses Remaining')
        elif guess > target + 1:
            print("Too High")
            print(f'{count} Guesses Remaining')
        else:
            print("Too Low")
            print(f'{count} Guesses Remaining')
    else:
        print("No Remaining Guesses")
        break
````

### Exercise 3 — Cookie Monster Game 🍪
Start with a number of cookies (read with `input()`). Each pass, the monster eats a **random** 1-3 cookies. Keep looping until no cookies remain, then print how many **bites** it took.

Example output:
```
3
1
2
2
It took 4 bites to eat all the cookies.
```
*Tips: `import random`; each pass `bite = random.randint(1, 3)`; loop `while cookies > 0`, subtract the bite, and count the bites. (The last bite may be bigger than what's left — that's fine, cookies just hit 0 or below.)*

````python
import random
# Your code here
cookies = int(input("Please input number of cookies:"))
tot_bite = 0
while cookies > 0:
    bite = random.randint(1,3)
    cookies -= bite
    tot_bite += 1
    print(bite)
print(f' It took {tot_bite} bites to eat all the cookies.')
    
````

### Exercise 4 — Design your own menu program 🏧
Build a menu-driven program with a `while` loop that keeps running until the user chooses **Exit**. The classic version is an **ATM** with `balance`, `deposit`, `withdraw`, and `Exit` — try building it yourself first.

*Note: Example 5 in the lecture is a worked ATM. Write yours from scratch, then compare — and if you finish early, **extend** it (e.g., add a 4-digit PIN check, a daily withdrawal limit, or a transaction history list).*

*Tips: `while True:` around the menu; read the choice with `input()`; handle each option with `if`/`elif`; `break` on Exit. Keep `balance` in a variable that persists across passes.*

````python
balance = 1000
# Your code here
PIN = 5876
while True:
    print("\nATM Menu:")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withrdaw")
    print("4. Exit")
    sel = int(input("Enter Selection:"))
    if sel == 1 or sel == 2 or sel == 3:
        attempt = int(input("Input PIN:"))
        if attempt != PIN:
            print("Access Denied")
        elif sel == 1 and attempt == PIN:
            print(balance)
        elif sel == 2 and attempt == PIN:
            dep = int(input("Deposit Amount:"))
            balance += dep
            print(f'Updated Balance:{balance}')
        elif sel == 3 and attempt == PIN:
            withdraw = int(input("Withdrawl Amount:"))
            balance -= withdraw
            print(f'Updated Balance:{balance}')
    elif sel == 4:
        break
````
